```python
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import SGDRegressor
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=4)
```

USing SDG Regression

```python
reg = SGDRegressor(penalty='l2',max_iter=500,eta0=0.1,learning_rate='constant',alpha=0.001)
```

```python
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
print('R2_Score : ',r2_score(y_test,y_pred))
print(reg.coef_)
print(reg.intercept_)
```

Using Ridge

```python
from sklearn.linear_model import Ridge
reg = Ridge(alpha=0.01,max_iter=500,solver='sparse_cg')
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)

print("R2_score : ",r2_score(y_test,y_pred))
print(reg.coef_)
print(reg.intercept_)
```

Creating Ridge Regression from Scratch

```python
class MyRidgeGD:
  def __init__(self,epochs,learning_rate=0.1,alpha=0.001):
    self.learning_rate = learning_rate
    self.epochs = epochs
    self.alpha = alpha
    self.coef_ = None
    self.intercept_ = None
  def fit(self,X_train,y_train):
    self.coef_ = np.ones(X_train.shape[1])
    self.intercept_ = 0
    theta = np.insert(self.coef_,0,self.intercept_)
    X_train = np.insert(X_train,0,1,axis=1)

    for i in range(self.epochs):
      theta_der = np.dot(X_train.T,(np.dot(X_train,theta) - y_train)) + self.alpha * theta
      theta = theta - self.learning_rate * theta_der
    self.coef_ = theta[1:]
    self.intercept_ = theta[0]
  def predict(self,X_test):
    return np.dot(X_test,self.coef_) + self.intercept_
```

```python
reg = MyRidgeGD(epochs=500,learning_rate=0.005,alpha=0.01)
```

```python
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
print('R2_Score : ',r2_score(y_test,y_pred))
print(reg.coef_)
print(reg.intercept_)
```
