```python
import pandas as pd
import numpy as np
from pandas_datareader import data
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn.metrics import r2_score
from sklearn.model_selection import cross_val_score,GridSearchCV
```

```python
from sklearn.datasets import fetch_california_housing
housing = fetch_california_housing()
data = housing.data
target = housing.target
```

```python
data.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(data,target,test_size=0.2,random_state=43)
```

```python
rt = DecisionTreeRegressor(criterion = 'squared_error' , max_depth=10)
```

```python
rt.fit(X_train,y_train)
```

```python
y_pred = rt.predict(X_test)
r2_score(y_test,y_pred)
```

# **Hyperparameter Tuning**

```python
param_grid = {
    'max_depth':[2,4,8,10,None],
    'criterion' : ['squared_error','absolute_error'],
    'max_features' : [0.25,0.5,1.0],
    'min_samples_split' : [0.25,0.5,1.0]
}
```

```python
reg = GridSearchCV(DecisionTreeRegressor(),param_grid=param_grid)
reg.fit(X_train,y_train)
```

```python
reg.best_score_
```

```python
reg.best_params_
```

# **Feature Importance**

```python
X_train_df = pd.DataFrame(X_train, columns=housing.feature_names)

# Now you can access the column names using X_train_df.columns
for importance, name in sorted(zip(rt.feature_importances_, X_train_df.columns), reverse=True):
    print(name, importance)
```
